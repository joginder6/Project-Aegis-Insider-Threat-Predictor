```python
import pandas as pd
import numpy as np

# 1. LOAD DATA
psych = pd.read_csv('E:\Internet Downloads\CERT.csv\psychometric.csv')
logon = pd.read_csv('E:\Internet Downloads\CERT.csv\logon.csv')
device = pd.read_csv('E:\Internet Downloads\CERT.csv\device.csv')
emails = pd.read_csv('E:\Internet Downloads\CERT.csv\email.csv')

# 2. CREATE A SHARED POOL OF IDs
# We take all unique users from BOTH files to make sure nobody is left out
all_activity_users = list(set(logon['user'].unique()) | set(emails['user'].unique()))

# 3. THE BRIDGE (Mapping)
# Assign a random "Activity ID" to every Psych student
np.random.seed(42)
psych['mapped_id'] = np.random.choice(all_activity_users, size=len(psych))

# 4. SUMMARIZE LOGONS (Night Activity)
logon['date'] = pd.to_datetime(logon['date'])
logon_sum = logon.groupby('user').agg(
    night_logons=('date', lambda x: ((x.dt.hour >= 23) | (x.dt.hour <= 5)).sum())
).reset_index()

# 5. SUMMARIZE EMAILS (Total Size)
# We use 'size' because that's the column in your email data
email_sum = emails.groupby('user')['size'].sum().reset_index(name='total_email_size')

# 6. SUMMARIZE DEVICES (USB Connects)
usb_sum = device[device['activity'] == 'Connect'].groupby('user').size().reset_index(name='usb_count')

# 7. THE MASTER MERGE
# We glue everything to 'psych' using the 'mapped_id' we created
master = psych.merge(logon_sum, left_on='mapped_id', right_on='user', how='left')
master = master.merge(email_sum, left_on='mapped_id', right_on='user', how='left')
master = master.merge(usb_sum, left_on='mapped_id', right_on='user', how='left')

# 8. CLEANUP
master.fillna(0, inplace=True)
# Remove the extra 'user' columns that pandas adds during a merge
master = master.loc[:, ~master.columns.duplicated()].copy()
cols_to_drop = [c for c in ['user', 'user_x', 'user_y'] if c in master.columns]
master.drop(columns=cols_to_drop, inplace=True)

# 9. FINAL CHECK
print(f"✅ Success! Rows with Night Logons: {len(master[master['night_logons'] > 0])}")
print(f"✅ Success! Rows with Email Data: {len(master[master['total_email_size'] > 0])}")
master.head()
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import xgboost as xgb
import joblib

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report, f1_score, precision_score, recall_score
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import AdaBoostClassifier
```

```python
df=pd.read_csv("cert_Data.csv")
df
```

```python
# defining what target value means...what criteria should be met !
df['target'] = (
    ((df['N'] > 35) | (df['C'] > 38)) & 
    ((df['night_logons'] > 0) | (df['usb_count'] > 0))
).astype(int)
```

```python
# data splitting (feature selection)
features = ['O','C','E','A','N','night_logons','total_email_size','usb_count']
x = df[features]
y = df['target']
```

```python
print("Total Safe Profiles in Dataset (0):", np.sum(y == 0))
print("Total Threat Profiles in Dataset (1):", np.sum(y == 1))
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y, test_size=0.2, random_state=42)
```

```python
scaler = StandardScaler()
x_train_scaled = pd.DataFrame(scaler.fit_transform(x_train), columns=features)
x_test_scaled = pd.DataFrame(scaler.transform(x_test), columns=features)
```

# ADABOOST Classifier

```python
param_grid={
'n_estimators':[50,100,150,200,250,300],
}
```

```python
weak_learner=DecisionTreeClassifier(max_depth=1)
```

```python

ad=AdaBoostClassifier(estimator=weak_learner,n_estimators=50, learning_rate=0.01, random_state=42)

grid_search=GridSearchCV(
    estimator=ad,
    param_grid=param_grid,
    cv=5,
    scoring='f1',
    verbose=1,
    n_jobs=-1
)
```

```python
print("Starting Grid Search CV....(this may take some time)")
grid_search.fit(x_train_scaled,y_train)
```

```python
best_grid_model = grid_search.best_estimator_
y_pred_grid = best_grid_model.predict(x_test_scaled)
```

```python
ada_accuracy = accuracy_score(y_test, y_pred_grid)
ada_f1 = f1_score(y_test, y_pred_grid, zero_division=0)
ada_recall = recall_score(y_test, y_pred_grid, zero_division=0)
```

```python
# 📊 LINE GRAPH FOR ADABOOST: GRID SEARCH ACCURACY PROGRESSION
print("Generating AdaBoost GridSearch Optimization Line Graph...")
cv_results = pd.DataFrame(grid_search.cv_results_)
estimators = cv_results['param_n_estimators'].astype(int)
mean_scores = cv_results['mean_test_score']
```

```python
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(estimators, mean_scores, marker='o', linestyle='-', color='#1f77b4', linewidth=2)
ax.axvline(x=grid_search.best_params_['n_estimators'], color='red', linestyle='--', alpha=0.7, label=f"Best Optima ({grid_search.best_params_['n_estimators']})")
ax.set_title('AdaBoost: Accuracy Curve vs. Estimator Volume')
ax.set_xlabel('Number of Estimators (n_estimators)')
ax.set_ylabel('Mean CV Validation Accuracy Score')
ax.grid(True, linestyle=':', alpha=0.6)
ax.legend()
plt.tight_layout()
plt.savefig('adaboost_learning_line.png', dpi=300)
plt.show()
```

```python
ada_accuracy = accuracy_score(y_test, y_pred_grid)
ada_f1 = f1_score(y_test, y_pred_grid, zero_division=0)
ada_recall = recall_score(y_test, y_pred_grid, zero_division=0)

print(f"🥇 AdaBoost Model Final Test Accuracy : {ada_accuracy * 100:.2f}%")
print(f"🎯 AdaBoost Model Threat Catch Rate (Recall): {ada_recall * 100:.2f}%")
print(f"⚖️  AdaBoost Model Balanced F1-Score      : {ada_f1 * 100:.2f}%")
print("-" * 60)
```

# XGBOOST CLassifier

```python
print("\n--- Training Standalone XGBoost Model ---")
ratio = (y_train == 0).sum() / (y_train == 1).sum()
optimal_trees = grid_search.best_params_['n_estimators']

xgb_model = xgb.XGBClassifier(
    n_estimators=optimal_trees, 
    scale_pos_weight=ratio,
    random_state=42,
    eval_metric='logloss'
)
```

```python
# Use an evaluation set during fit to capture the real iteration training error
xgb_model.fit(
    x_train_scaled, y_train,
    eval_set=[(x_train_scaled, y_train), (x_test_scaled, y_test)],
    verbose=False
)
```

```python
xgb_preds = xgb_model.predict(x_test_scaled)
```

```python
xgb_accuracy = accuracy_score(y_test, xgb_preds)
xgb_f1 = f1_score(y_test, xgb_preds, zero_division=0)
xgb_recall = recall_score(y_test, xgb_preds, zero_division=0)
```

```python
# 📊 LINE GRAPH FOR XGBOOST: LOSS DROP OVER ITERATIONS
print("Generating XGBoost Loss Optimization Line Graph...")
eval_results = xgb_model.evals_result()
train_loss = eval_results['validation_0']['logloss']
test_loss = eval_results['validation_1']['logloss']

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(train_loss, label='Train Optimization Loss', color='blue', alpha=0.7)
ax.plot(test_loss, label='Test Validation Loss', color='orange', alpha=0.7)
ax.set_title('XGBoost: Sequential Error Loss Optimization Curve')
ax.set_xlabel('Boosting Iterations / Tree Splits')
ax.set_ylabel('Logarithmic Log-Loss Error')
ax.grid(True, linestyle=':', alpha=0.6)
ax.legend()
plt.tight_layout()
plt.savefig('xgboost_learning_line.png', dpi=300)
plt.show()
```

```python
xgb_accuracy = accuracy_score(y_test, xgb_preds)
print(f"⚡ XGBoost Model Final Test Accuracy Score: {xgb_accuracy * 100:.2f}%")
print("-" * 60)
```

```python
# Graph A: AdaBoost Hyperparameter Curve
cv_results = pd.DataFrame(grid_search.cv_results_)
estimators = cv_results['param_n_estimators'].astype(int)
mean_scores = cv_results['mean_test_score']

fig, ax = plt.subplots(figsize=(6, 3.5))
ax.plot(estimators, mean_scores, marker='o', linestyle='-', color='#1f77b4', linewidth=2)
ax.axvline(x=grid_search.best_params_['n_estimators'], color='red', linestyle='--', alpha=0.7, label='Best Optima')
ax.set_title('AdaBoost: F1 Optimization Curve')
ax.set_xlabel('Number of Estimators')
ax.set_ylabel('Mean CV F1 Score')
ax.grid(True, linestyle=':', alpha=0.6)
plt.tight_layout()
plt.savefig('adaboost_learning_line.png', dpi=300)
plt.show()
```

```python
# Graph B: XGBoost Loss Drop Curve
eval_results = xgb_model.evals_result()
train_loss = eval_results['validation_0']['logloss']
test_loss = eval_results['validation_1']['logloss']

fig, ax = plt.subplots(figsize=(6, 3.5))
ax.plot(train_loss, label='Train Loss', color='blue', alpha=0.7)
ax.plot(test_loss, label='Test Loss', color='orange', alpha=0.7)
ax.set_title('XGBoost: Loss Optimization Curve')
ax.set_xlabel('Iterations')
ax.set_ylabel('Log-Loss Error')
ax.grid(True, linestyle=':', alpha=0.6)
ax.legend()
plt.tight_layout()
plt.savefig('xgboost_learning_line.png', dpi=300)
plt.show()
```

```python
# NEW Graph C: Model Performance Engine Comparison Chart
metrics_df = pd.DataFrame({
    'Metric': ['Accuracy', 'Recall (Catch Rate)', 'F1-Score'] * 2,
    'Score': [ada_accuracy, ada_recall, ada_f1, xgb_accuracy, xgb_recall, xgb_f1],
    'Engine Model': ['AdaBoost', 'AdaBoost', 'AdaBoost', 'XGBoost', 'XGBoost', 'XGBoost']
})

plt.figure(figsize=(7, 4.5))
sns.barplot(x='Metric', y='Score', hue='Engine Model', data=metrics_df, palette='Set2')
plt.title('Benchmark Matrix: AdaBoost vs. XGBoost Protocol')
plt.ylim(0, 1.05)
plt.ylabel('Performance Rating Value')
for p in plt.gca().patches:
    if p.get_height() > 0:
        plt.gca().annotate(f"{p.get_height()*100:.1f}%", (p.get_x() + p.get_width() / 2., p.get_height() - 0.08),
                    ha='center', va='center', color='white', fontweight='bold', xytext=(0, 0), textcoords='offset points')
plt.tight_layout()
plt.savefig('model_comparison_bar.png', dpi=300)
plt.show()
```

```python
# 5. LIVE SINGLE TARGET USER EVALUATION WITH HYBRID EXPERT RULE OVERRIDE
# =====================================================================
print("\n--- Live Single Target User Evaluation ---")
o = float(input("Enter O: "))
c = float(input("Enter C: "))
e = float(input("Enter E: "))
a = float(input("Enter A: "))
n = float(input("Enter N: "))
logons = float(input("Enter Night Logons: "))
usb = float(input("Enter USB Count: "))
email = float(input("Enter Email Size: "))

person_dict = {
    "O": o, "C": c, "E": e, "A": a, "N": n,
    "night_logons": logons, "total_email_size": email, "usb_count": usb
}
person_df = pd.DataFrame([person_dict])
person_scaled = pd.DataFrame(scaler.transform(person_df), columns=features)

print("\n🤖 CHOOSE YOUR DETECTION ENGINE PROTOCOL:")
print("1. AdaBoost (Optimized GridSearch Classifier)")
print("2. XGBoost (Extreme Gradient Boosting Signature Classifier)")

choice = input("Enter your engine selection choice (1 or 2): ").strip()

# --- HYBRID EXPERT RULE CORRECTION OVERRIDE LAYER ---
# Triggers an immediate risk flag if the exact threshold triggers are matched
is_expert_rule_breached = (logons > 1 and usb > 2 and a > 38 and n > 38)

print("\n" + "="*45)
if choice == '1':
    print("Executing Protocol: AdaBoost Grid Engine Matrix...")
    if is_expert_rule_breached:
        result = 1
    else:
        result = best_grid_model.predict(person_scaled)[0]
        
elif choice == '2':
    print("Executing Protocol: XGBoost Targeted Balancing Engine...")
    if is_expert_rule_breached:
        result = 1
    else:
        result = xgb_model.predict(person_scaled)[0]
        
else:
    print("⚠️ Unknown selection command! Defaulting to Core AdaBoost...")
    result = 1 if is_expert_rule_breached else best_grid_model.predict(person_scaled)[0]

print(f"Prediction Output Metric Choice: {result}")
if result == 1:
    print("Final Assessment Status: 🚩 THREAT BOUNDARY DEVIATION DETECTED")
else:
    print("Final Assessment Status: ✅ SAFE BEHAVIOR PROFILE CONFIRMED")
print("="*45)
```

```python
# =====================================================================
# 6. ENTERPRISE BATCH LOGGING SCAN
# =====================================================================
print("\n--- Running Enterprise Batch Logging Scan ---")
new_data = pd.read_csv("cert_Data.csv")
x_new = new_data[features]
x_new_scaled = pd.DataFrame(scaler.transform(x_new), columns=features)

predictions = best_grid_model.predict(x_new_scaled)
probabilities = best_grid_model.predict_proba(x_new_scaled)[:, 1]

new_data['Risk_Score'] = (probabilities * 100).round(2)
new_data['Final_Status'] = ["🚩 THREAT" if p == 1 else "✅ SAFE" for p in predictions]

# Save full results to excel
new_data.to_excel("Batch_Report.xlsx", index=False)
print("✅ Batch Prediction Complete! Results saved to Batch_Report.xlsx")

# --- LIVE BATCH REPORT SUMMARY FOR VIVA PRESENTATION ---
total_scanned = len(new_data)
total_threats = np.sum(predictions == 1)
threat_percentage = (total_threats / total_scanned) * 100

print("\n" + "="*60)
print("             🚨 BATCH SCAN EXECUTIVE SUMMARY            ")
print("="*60)
print(f"Total Employee Log Profiles Scanned : {total_scanned}")
print(f"Malicious Threat Profiles Identified: {total_threats} ({threat_percentage:.2f}%)")
print("="*60)

if total_threats > 0:
    print(f"\n📋 DETAILED LIST OF ALL {total_threats} FLAGGED THREAT PROFILES:")
    
    # Filter out only the threats and sort them by the highest risk score first
    threat_profiles = new_data[new_data['Final_Status'] == "🚩 THREAT"].sort_values(by='Risk_Score', ascending=False).copy()
    
    # Automatically look for an identifying column in your CSV
    identity_col = None
    for col in ['Name', 'Employee_ID', 'User', 'id', 'employee', 'username']:
        if col in new_data.columns:
            identity_col = col
            break
            
    # Dynamic column layout based on whether a identifier column exists
    if identity_col:
        # If names/IDs exist, show them first
        display_cols = [identity_col, 'Risk_Score', 'N', 'C', 'night_logons', 'usb_count']
        print(threat_profiles[display_cols].to_string(index=False))
    else:
        # If no identity column exists, create a temporary "Row_Number" so you can find them in Excel
        threat_profiles['Excel_Row'] = threat_profiles.index + 2  # +2 accounts for 0-index and header row in Excel
        display_cols = ['Excel_Row', 'Risk_Score', 'N', 'C', 'night_logons', 'usb_count']
        print("Note: No explicit Name/ID column found. Displaying matching Excel row numbers:")
        print("-" * 60)
        print(threat_profiles[display_cols].to_string(index=False))
else:
    print("\n✅ Zero anomalies detected across the enterprise batch registry.")
print("="*60 + "\n")
```
